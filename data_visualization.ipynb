```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1,2,3,4]
y = [10,15,13,20]

plt.plot(x, y)
plt.show()
```

```python
plt.bar(["A", "B", "C"], [5, 8, 3])
plt.show()
```

```python
data = np.random.randn(1000)

plt.hist(data, bins=30)
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,5,4,5]

plt.scatter(x, y)
plt.show()
```

```python
plt.pie([30, 50, 20], labels=["A", "B", "C"])
plt.show()
```
